```python
# Importing libraries
import os
import json
from dotenv import load_dotenv
from openai import OpenAI, AzureOpenAI
from anthropic import AnthropicFoundry
from IPython.display import Markdown,display
```

```python
# let's load the keys/variables
load_dotenv(override=True)  
```

```python
# Let's see if the API key is working/helping us to call LLM from Azure Foundry
import os
# From OpenAI
AZURE_OPENAI_API_KEY= os.getenv("AZURE_OPENAI_API_KEY")
AZURE_OPENAI_MODEL_DEPLOYMENT = os.getenv("AZURE_OPENAI_DEPLOYMENT")
AZURE_OPENAI_ENDPOINT = os.getenv("AZURE_OPENAI_ENDPOINT")
AZURE_OPENAI_API_VERSION = os.getenv("AZURE_OPENAI_API_VERSION")
AZURE_OPENAI_DEPLOYMENT_GPT_41 = os.getenv("AZURE_OPENAI_DEPLOYMENT_GPT_41")
AZURE_OPENAI_DEPLOYMENT_GPT_54_mini = os.getenv("AZURE_OPENAI_DEPLOYMENT_GPT_54_mini")
AZURE_OPENAI_DEPLOYMENT_GPT_55 = os.getenv("AZURE_OPENAI_DEPLOYMENT_GPT_55")
AZURE_OPENAI_DEPLOYMENT_GPT_4O_mini = os.getenv("AZURE_OPENAI_DEPLOYMENT_GPT_4O_mini")
if AZURE_OPENAI_API_KEY:
    print("AZURE_OPENAI_API_KEY is available")
else:
    print("AZURE_OPENAI_API_KEY is not available")

# From Anthropic
AZURE_CLAUDE_DEPLOYMENT_OPUS_48=os.getenv("AZURE_CLAUDE_DEPLOYMENT_OPUS_48")
AZURE_CLAUDE_ENDPOINT=os.getenv("AZURE_CLAUDE_ENDPOINT")
AZURE_CLAUDE_API_KEY=os.getenv("AZURE_CLAUDE_API_KEY")
if AZURE_CLAUDE_API_KEY:
    print("AZURE_CLAUDE_API_KEY is avaiable")
else:
    print("AZURE_CLAUDE_API_KEY is not available")

```

```python
# Set the client for AzureOpenAI LLM callings
client_OpenAI = AzureOpenAI(
    api_version=AZURE_OPENAI_API_VERSION,
    azure_endpoint=AZURE_OPENAI_ENDPOINT,
    api_key=AZURE_OPENAI_API_KEY
)
```

```python
# Set the client for Anthropic model
client_anthropic=AnthropicFoundry(
    api_key=AZURE_CLAUDE_API_KEY,
    base_url=AZURE_CLAUDE_ENDPOINT
)
```

```python
# Let's set a request for the AI Models
request = """
    Please come up with a challenging, nuanced question with a succint answer,
    that I can ask a number of LLMs to evaluate their intelligence. 
    Not a mathematical puzzle, but more of a thought-provoking question which required intelligent insight.
    Include in your question that the answer must be short.
"""

request += "Answer only wuth question, no explanation."
messages = [
    {
        "role": "user",
        "content": request
    }
]
print(messages)
```

```python
# Let's get a response from a best AI Model (gpt-5.5)
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_55,
    messages=messages
)
question = response.choices[0].message.content
print(f"Question from {AZURE_OPENAI_DEPLOYMENT_GPT_55} AI Model is: \n", question)
```

```python
# let's prepare the message for all other models
competitors=[]
answers=[]
messages=[
    {
        "role":"user",
        "content":question
    }
]
print(messages)
```

```python
# create a record function for recording the answer of the models
def record(model_name, answer):
    competitors.append(model_name)
    answers.append(answer)
    print(f"Answer have been recorded from {model_name} AI model.")
    print("Model's answer: \n")
    display(Markdown(answer))
```

```python
# Let's get a response from a Model (gpt-4.1)
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_41,
    messages=messages
)
answer = response.choices[0].message.content
record(AZURE_OPENAI_DEPLOYMENT_GPT_41,answer)
```

```python
# Let's get a response from a Model (gpt-4o-mini)
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_4O_mini,
    messages=messages
)
answer = response.choices[0].message.content
record(AZURE_OPENAI_DEPLOYMENT_GPT_4O_mini,answer)
```

```python
# Let's get a response from a Model (gpt-5.4-mini)
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_54_mini,
    messages=messages
)
answer = response.choices[0].message.content
record(AZURE_OPENAI_DEPLOYMENT_GPT_54_mini,answer)
```

```python
# Let's get a response from a Model (gpt-5.5)
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_55,
    messages=messages
)
answer = response.choices[0].message.content
record(AZURE_OPENAI_DEPLOYMENT_GPT_55,answer)
```

```python
# Let's get a response from a model (Claude-Opus-4.8)

response = client_anthropic.messages.create(
    model=AZURE_CLAUDE_DEPLOYMENT_OPUS_48,
    messages=messages,
    max_tokens=2000
)
answer = response.content[0].text
record(AZURE_CLAUDE_DEPLOYMENT_OPUS_48,answer)
```

```python
# Let's print all the models answer
print(f"Number of models answered the question: {len(competitors)}")
print("Models: ",competitors)
print("Models answer: ", answers)
```

```python
# let's zip the results
print(f"Question: {question}\n")
for competitor, answer in zip(competitors,answers):
    print(f"competitors model name: {competitor}\nAnswer: {answer}")
```

```python
# Let's bring this together - note the use of "enumerate"
together = ""
for index, answer in enumerate(answers):
    together += f"## Response from AI Model {index + 1}\n"
    together += answer + "\n\n"
```

```python
print(together)
```

```python
# Let's prepare the prompt for an AI model to judge all the answers for the question
judge = f"""
You are judging a competition between {len(competitors)} competitor AI models.
Each model has been given this question:
{question}

Your job is to evaluate each response from clarity and strength of argument, and rank them in order of best to worst.
Respond with .json format and only .json, with the following format:
{{"results": ["best competitor number", "second best competitor number", "third best competitor number", and so on]}}

Here is the responses for each of the competitor model:

{together}

Now respond with .json format with the ranked order of the competitors, nothing else, Do not include markdown formatting  or any kind of other textual information.
"""
```

```python
print(judge)
```

```python
judge_messages = [
    {
        "role": "user",
        "content":judge
    }
]
print(judge_messages)
```

```python
# Let's use claude opus 4.8 as a LLM judge for all the answer even though there is one answer from his side also included.
response = client_anthropic.messages.create(
    model=AZURE_CLAUDE_DEPLOYMENT_OPUS_48,
    messages=judge_messages,
    max_tokens=10000
)
```

```python
# Let's use gpt-5.5 as a LLM judge for all the answer even though there is one answer from his side also included.
response = client_OpenAI.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_GPT_55,
    messages=judge_messages
)
answer_gpt_55 = response.choices[0].message.content
print(f"AI model {AZURE_OPENAI_DEPLOYMENT_GPT_55} judged all the answers in below presented way: \n {answer_gpt_55}")
```

```python
answer_claude_OPUS_48 = response.content[0].text
print(f"AI model {AZURE_CLAUDE_DEPLOYMENT_OPUS_48} judged all the answers in below presented way: \n {answer_claude_OPUS_48}")
```

```python
# Let's print the final results from both judges
def print_results(results, used_model):
    results_dict = json.loads(results)
    ranks = results_dict["results"]
    print(f"Results judged by {used_model} AI model.")
    for index, result in enumerate(ranks):
        competitor = competitors[int(result) -1]
        print(f"Rank {index + 1}: {competitor}")
    print("\n")
```

```python
print_results(answer_gpt_55,AZURE_OPENAI_DEPLOYMENT_GPT_55)
print_results(answer_claude_OPUS_48, AZURE_CLAUDE_DEPLOYMENT_OPUS_48)
```
