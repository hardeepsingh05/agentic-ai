```python
# Let's see if the API key is working/helping us to call LLM from Azure Foundry
import os
AZURE_OPENAI_API_KEY= os.getenv("AZURE_OPENAI_API_KEY")
AZURE_OPENAI_MODEL_DEPLOYMENT = os.getenv("AZURE_OPENAI_DEPLOYMENT")
AZURE_OPENAI_ENDPOINT = os.getenv("AZURE_OPENAI_ENDPOINT")
AZURE_OPENAI_API_VERSION = os.getenv("AZURE_OPENAI_API_VERSION")
AZURE_OPENAI_MODEL_DEPLOYMENT_2 = os.getenv("AZURE_OPENAI_DEPLOYMENT_2")
if AZURE_OPENAI_API_KEY:
    print("API Key Available")
else:
    print("No Available")
```

```python
# Importing library
from openai import OpenAI, AzureOpenAI
```

```python
# Set the client for AzureOpenAI LLM callings
client = AzureOpenAI(
    api_version=AZURE_OPENAI_API_VERSION,
    azure_endpoint=AZURE_OPENAI_ENDPOINT,
    api_key=AZURE_OPENAI_API_KEY
)
```

```python
# create a list of message in the familiar OpenAI format
messages = [{
    "role": "user",
    "content": "Tell me a fun fact"
}]
```

```python
messages
```

```python
# Let's get a response from our model
response = client.chat.completions.create(
    messages=messages,
    model=AZURE_OPENAI_MODEL_DEPLOYMENT)
print(response.choices[0].message.content)
```

```python
# And now - let's ask some questions
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with question."
messages = [
    {
        "role": "user",
        "content":question
    }
]
question = client.chat.completions.create(
    messages=messages,
    model=AZURE_OPENAI_MODEL_DEPLOYMENT
)
print(question.choices[0].message.content)
```

```python
question = question.choices[0].message.content
print(question)
```

```python
messages = [
    {
        "role": "user",
        "content":question
    }
]
messages
```

```python
# Ask model itself to answer this hard question
answer = client.chat.completions.create(
    messages=messages,
    model=AZURE_OPENAI_MODEL_DEPLOYMENT
)
print(answer.choices[0].message.content)
```

```python
answer = answer.choices[0].message.content
```
